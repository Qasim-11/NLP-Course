# 1- TF-IDF

<img src="https://drive.google.com/uc?export=view&id=1saSt0mMgOQ2ybbTms1lu_ruhUcBWkKzL" width="500">

TF-IDF stands for term frequency-inverse document frequency. It is a measure that discounts common words. Used in the fields of information retrieval (IR) and machine learning, that can quantify the importance of words in a document amongst a collection of documents (also known as a corpus).

### Components of TF-IDF

1. **TF (Term Frequency):**  
   Measures how frequently a term appears in a document.


$$ \text{TF}(t, d) = \frac{\text{Number of times term } t \text{ appears in document } d}{\text{Total number of terms in document } d} $$


2. **IDF (Inverse Document Frequency):**  
   Measures how important a term is across all documents. Words that appear in many documents get lower scores.

$$    \text{IDF}(t) = \log \left(\frac{\text{Number of all documents N}}{\text{Number of documents containing the term } t}\right) $$

# Cosine similarity

In NLP, Cosine similarity is a metric used to measure how similar the documents are.


$$ \text{cosine similarity} = \frac{A \cdot B}{\|A\| \times \|B\|} $$

Where:  
$ A \cdot B $  = dot product of vectors A and B  
$ \|A\| $ = magnitude (length) of vector A  
$ \|B\| $ =  magnitude (length) of vector B

#### Intuition

- If vectors point in the **same direction**, cosine similarity = **1** (maximum similarity).
- If vectors are **orthogonal (90° apart)**, cosine similarity = **0** (no similarity).

Word2Vec utilizes two architectures :

````python
# Task 1: cosine similarity between the four sentences
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

sentences = [
    'This is the first document.',
    'This document is the second document.',
    'And this is the third one.',
    'Is this the first document?',
]

vectors = TfidfVectorizer().fit_transform(sentences)

pd.DataFrame(cosine_similarity(vectors),
             index=sentences, columns=sentences).round(3)
````

> **Your answers:**

>    notion of word order or punctuation, so a statement and a question look the same to it.
> 2. Sentences 1 and 2 score **0.647** - they share *this*, *is*, *the* and *document*.
> 3. Sentence 3 is the odd one out, at **0.308** and **0.225** against the first two. It is the only

### Task 2: TF-IDF
Use tf-idf method on the sentences below to determine the important words.

'data science is one of the most important fields of science',  
'this is one of the best data science courses',  
'data scientists analyze data'  


````python
# Task 2: tf-idf on the three sentences.
# The highest score in a row is that sentence's most important word.
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer

docs = [
    'data science is one of the most important fields of science',
    'this is one of the best data science courses',
    'data scientists analyze data',
]

tfidf = TfidfVectorizer()
result = tfidf.fit_transform(docs)

scores = pd.DataFrame(result.toarray(), columns=tfidf.get_feature_names_out())
display(scores.round(3))

for i, doc in enumerate(docs):
    top = scores.iloc[i].sort_values(ascending=False).head(3)
    print("\n" + doc)
    print("   -> " + ", ".join("%s (%.3f)" % (w, s) for w, s in top.items()))
````

>    short sentence. That discounting of common words is the whole point of TF-IDF.

````python
# Task 3: the Simpsons dataset -> clean it -> train a skip gram Word2Vec model
import os
import re
import pandas as pd
import gensim
from nltk.tokenize import word_tokenize

def find_file(name):
    for path in [name, os.path.join("dataset", name), os.path.join("/content", name)]:
        if os.path.exists(path):
            return path
    raise FileNotFoundError(name)

simpsons = pd.read_csv(find_file("simpsons_script_lines.csv"))
lines = simpsons["spoken_words"].dropna()

def clean_text(text):
    text = text.lower()
    text = re.sub(r"[0-9]", '', text)
    text = re.sub(r"[)(,”“’$-]", '', text)
    return text

tokens = [word_tokenize(clean_text(line)) for line in lines]
tokens = [t for t in tokens if t]

print("lines:", len(tokens))

Skip_gram_model = gensim.models.Word2Vec(tokens, min_count=1, vector_size=100,
                                         window=5, sg=1)
print("vocabulary:", len(Skip_gram_model.wv))
````

### Task 4

Use: wv.most_similar() method to :

1.	Find the words similar to “homer”.

2.	Find the words similar to “marge”.

3. Find the words similar to “bart”



````python
# Task 4: wv.most_similar()
for name in ["homer", "marge", "bart"]:
    print(name)
    for word, score in Skip_gram_model.wv.most_similar(name, topn=5):
        print("   %-15s %.3f" % (word, score))
    print()
````
